1. Charger le fichier Excel

```python
import pandas as pd
import numpy as np
chemin = "/content/drive/MyDrive/decodelab internship/Dataset for Data Analytics.xlsx"
df = pd.read_excel(chemin)
df.head()
```

2. Voir les informations du dataset

```python
print(df.shape)
print(df.dtypes)
print(df.isnull().sum())
```

3. Remplacer les valeurs vides

```python
df.replace(["", " ", "N/A", "NA", "null", "-", "?"], np.nan, inplace=True)
```

```python
for colonne in df.select_dtypes(include="number").columns:
    df[colonne] = df[colonne].fillna(df[colonne].median())
```

```python
df["CouponCode"] = df["CouponCode"].fillna("No Coupon")
for colonne in df.select_dtypes(include="object").columns:
    if colonne != "CouponCode" and df[colonne].isnull().any():
        df[colonne] = df[colonne].fillna(df[colonne].mode()[0])
```

4. Supprimer les doublons

```python
print("Doublons avant nettoyage :", df.duplicated().sum())

df = df.drop_duplicates()

print("Doublons après nettoyage :", df.duplicated().sum())
```

5. Supprimer la colonne "date" dupliquée

```python
if "date" in df.columns:
    df = df.drop(columns=["date"])
```

6. Nettoyer les textes

```python
for colonne in df.select_dtypes(include="object").columns:
    df[colonne] = df[colonne].str.strip()
```

7. Corriger les dates

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce",
    dayfirst=True
)
```

8. Corriger les nombres

```python
df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
).round(2)
```

```python
df["TotalPrice"] = pd.to_numeric(
    df["TotalPrice"],
    errors="coerce"
).round(2)
```

9. Vérification finale

```python
print("Valeurs manquantes :")
print(df.isnull().sum())

print("\nNombre de doublons :")
print(df.duplicated().sum())

print("\nTypes des colonnes :")
print(df.dtypes)
```

10. Enregistrer le fichier nettoyé

```python
sortie = "/content/drive/MyDrive/decodelab internship/Cleaned_Dataset.xlsx"

df.to_excel(sortie, index=False)

print("Fichier enregistré :", sortie)
```
